**Data Kepesertaan**

Untuk dataset Kepersertaan 2019-2021 library pandas diimpor untuk mengelola data, kemudian file Kepersertaan 2019-2021 dimuat ke dalam DataFrame dm. Dataset Kepersertaan 2019-2021 mencakup data kepesertaan JKN-KIS tahun 2019-2021, termasuk informasi penting seperti nomor peserta (identifikasi unik untuk setiap peserta), tanggal lahir, jenis kelamin, status perkawinan, hubungan keluarga, dan tempat tinggal (provinsi dan kabupaten/kota). Kolom-kolom ini memungkinkan analisis demografi peserta untuk memahami profil populasi yang terdaftar pada tahun tersebut.
Selanjutnya, menampilkan struktur data dm menggunakan dm.info(), yang memberikan informasi tentang jumlah baris, tipe data, serta keberadaan nilai kosong. Ini penting untuk memastikan bahwa data telah dikumpulkan dengan benar dan siap digunakan untuk analisis lebih lanjut.
Berikut penjelasan tiap kolom dari dataset Kepersertaan 2019-2021 : 
- PSTV01: Nomor Peserta – Identifikasi unik untuk peserta.
- PSTV02: Nomor kedua untuk identifikasi, mungkin terkait dengan keperluan internal atau tambahan identifikasi.
- PSTV03: Tanggal lahir peserta – Informasi tentang kapan peserta lahir.
- PSTV04: Hubungan dalam keluarga – Menunjukkan status peserta dalam keluarga, seperti "PESERTA", "ISTRI", "SUAMI".
- PSTV05: Jenis kelamin – Apakah peserta berjenis kelamin "LAKI-LAKI" atau "PEREMPUAN".
- PSTV06: Status perkawinan – Status pernikahan peserta seperti "KAWIN" atau "CERAI".
- PSTV07: Kelas premi – Menunjukkan tingkat layanan yang diikuti oleh peserta, seperti "KELAS I".
- PSTV08: Segmentasi peserta – Kategori peserta seperti "PPU" (Pekerja Penerima Upah), "BUKAN PEKERJA".
- PSTV09: Provinsi tempat tinggal – Provinsi di mana peserta tinggal, misalnya "ACEH", "SUMATERA UTARA".
- PSTV10: Kabupaten/Kota tempat tinggal – Menunjukkan kabupaten/kota peserta, seperti "PIDIE", "KOTA BANDA ACEH".
- PSTV11: Pemerintah penyedia layanan – Informasi tentang pihak penyedia, misalnya "PEMERINTAH KABUPATEN/KOTA".
- PSTV12: Jenis fasilitas kesehatan – Contohnya "PUSKESMAS".
- PSTV13: Provinsi fasilitas kesehatan – Provinsi tempat faskes tersebut berada.
- PSTV14: Kabupaten/Kota fasilitas kesehatan – Informasi kabupaten/kota dari faskes.
- PSTV15: Bobot – Angka yang menunjukkan bobot data untuk keperluan analisis statistik.
- PSTV16: Tahun data terakhir diperbarui – Tahun terakhir di mana data peserta diperbarui, seperti "2020".
- PSTV17: Status aktif peserta – Menunjukkan apakah peserta masih "AKTIF" atau "MENINGGAL".
- PSTV18: Tahun tertentu yang mungkin menunjukkan tahun kejadian atau status tertentu (kolom ini mungkin berisi data numerik atau bisa kosong).

- FKP15A: Informasi tambahan tentang klasifikasi layanan atau prosedur medis.

- FKP18: Kode yang bisa menunjukkan jenis tindakan medis tertentu.
- FKP19: Angka yang menggambarkan detail tambahan dalam prosedur atau layanan medis.
- FKP20: Indikator terkait jenis layanan yang diberikan kepada peserta.
- FKP21: Kode atau angka yang terkait dengan klasifikasi layanan lanjutan.
