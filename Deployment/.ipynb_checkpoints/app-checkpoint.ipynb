```python
import numpy as np
from flask import Flask, request, render_template
import dill
import nest_asyncio
import traceback

# Terapkan nest_asyncio
nest_asyncio.apply()

# Buat aplikasi Flask
app = Flask(__name__)

# Global variable untuk model
model = None

# Fungsi untuk memuat model dengan error handling yang lebih baik
def load_model():
    global model
    try:
        with open("mdl.pkl", "rb") as file:
            model = dill.load(file)
        print("Model berhasil dimuat!")
    except FileNotFoundError:
        print("Model file 'mdl.pkl' tidak ditemukan. Pastikan file ada di direktori yang benar.")
        raise
    except Exception as e:
        print(f"Gagal memuat model: {e}")
        print(traceback.format_exc())
        raise

# Mapping kelas status kunjungan
status_kunjungan_mapping = {
    0: 'Berobat jalan',
    1: 'Rujuk lanjut', 
    2: 'Sembuh',
    3: 'Kunjungan sehat',
    4: 'Lain-lain/Pulang paksa/Meninggal'
}

# Route untuk halaman utama
@app.route("/")
def home():
    return render_template("index.html")

# Route untuk prediksi
@app.route("/predict", methods=["POST"])
def predict():
    if model is None:
        return render_template("index.html", prediction_text="Model belum dimuat. Hubungi administrator.")

    try:
        # Validasi dan konversi input
        input_features = [
            int(request.form['Kode_Provinsi']),
            int(request.form['Kode_Jenis_Poliklinik']),
            int(request.form['Kode_Diagnosis']),
            int(request.form['Kode_Jenis_Kunjungan'])
        ]

        # Pastikan jumlah fitur sesuai
        if len(input_features) != 4:
            return render_template("index.html", prediction_text="Harus memasukkan 4 fitur: Kode Provinsi, Kode Jenis Poliklinik, Kode Diagnosis, Kode Jenis Kunjungan")

        # Konversi ke array numpy
        features = np.array([input_features])

        # Prediksi probabilitas
        probabilities = model.predict_proba(features)

        # Hitung jumlah pasien untuk setiap kelas (dalam bentuk prosentase)
        jumlah_pasien = {status_kunjungan_mapping[i]: round(probabilities[0][i] * 100) for i in range(len(probabilities[0]))}

        # Menyiapkan output prediksi dalam bentuk tabel
        return render_template("index.html", prediction_results=jumlah_pasien)

    except ValueError:
        return render_template("index.html", prediction_text="Input tidak valid. Mohon masukkan nilai numerik.")
    except Exception as e:
        # Log error untuk debugging
        print(f"Error dalam prediksi: {str(e)}")
        print(traceback.format_exc())
        return render_template("index.html", prediction_text=f"Terjadi kesalahan: {str(e)}")

# Muat model sebelum menjalankan server
load_model()

# Jalankan aplikasi
if __name__ == "__main__":
    try:
        app.run(debug=True, host='0.0.0.0', port=5000, use_reloader=False)
    except Exception as e:
        print(f"Gagal menjalankan server: {e}")
        print(traceback.format_exc())
```
